## Question 16 — Student Marks Processor

**Problem:**  
A teacher has a list of marks of all students in a class and wants key statistics.

**Requirements:**
- Create a function that accepts a **list of student marks**.
- Using a loop (or appropriate logic), calculate and **return**:
  - Highest mark
  - Lowest mark
  - Total marks
  - Average marks
- Display the results outside the function.

**Focus:** Function + list + loop + `return`

````python
# Write your code below

def marks_statistics(marks):
    if len(marks) == 0:
        return None, None, 0, 0
    highest = marks[0]
    lowest = marks[0]
    total = 0

    for mark in marks:
        if mark > highest:
            highest = mark
        if mark < lowest:
            lowest = mark
        total += mark

    average = total / len(marks)
    return highest, lowest, total, average

highest, lowest, total, average = marks_statistics([78, 90, 65, 88, 72])
print("Highest:", highest)
print("Lowest:", lowest)
print("Total:", total)
print("Average:", average)
````

## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
# Write your code below

def process_order(price, quantity, discount_percent, tax_percent):
    def calculate_subtotal():
        return price * quantity

    def calculate_discount(subtotal):
        return subtotal * discount_percent / 100

    def calculate_tax(amount):
        return amount * tax_percent / 100

    def calculate_final_amount():
        subtotal = calculate_subtotal()
        after_discount = subtotal - calculate_discount(subtotal)
        return after_discount + calculate_tax(after_discount)

    return calculate_final_amount()

amount = process_order(1000, 3, 10, 5)
print("Final Order Amount: Rs.", amount)
````

## Question 27 — Employee Payroll System

**Problem:**  
Build a small employee payroll system using multiple functions.

**Requirements:**
Your program should contain **separate functions** for:
- Employee information
- Basic salary

- Deduction

**Rules:**
- Use arguments and `return` properly.
- Use conditions where required (e.g., tax slabs).

- Display a clean final salary report.

**Focus:** Function design + modular programming

````python
# Write your code below
def employee_info(name, employee_id):
    return {"Name": name, "Employee ID": employee_id}

def basic_salary(amount):
    return amount

def allowance(amount):
    return amount * 0.10

def deduction(amount):
    return amount * 0.05

def tax(amount):
    if amount > 100000:
        return amount * 0.10
    elif amount > 50000:
        return amount * 0.05
    else:
        return 0

def calculate_final_salary(basic):
    gross = basic + allowance(basic) - deduction(basic)
    tax_amount = tax(gross)
    return gross - tax_amount, gross, tax_amount

employee = employee_info("Hamza", "EMP101")
basic = basic_salary(60000)
final, gross, tax_amount = calculate_final_salary(basic)

print("----- EMPLOYEE PAYROLL -----")
for key, value in employee.items():
    print(f"{key}: {value}")
print("Basic Salary: Rs.", basic)
print("Gross Salary: Rs.", gross)
print("Tax: Rs.", tax_amount)
print("Final Salary: Rs.", final)
````

## Question 28 — Smart Shopping Billing System

**Problem:**  
Build a shopping billing system that can handle multiple products.

**Requirements:**

- Calculate:
  - Individual product totals
  - Overall subtotal

  - Final bill
- Use **multiple functions** to separate responsibilities.
- At least one function must use `*args` or a flexible-argument design.

````python
# Write your code below

def product_total(name, price, quantity):
    total = price * quantity
    print(f"{name}: {quantity} x Rs. {price} = Rs. {total}")
    return total

def subtotal(*totals):
    return sum(totals)

def calculate_discount(amount, percent=10):
    return amount * percent / 100

def calculate_tax(amount, percent=5):
    return amount * percent / 100

items = [
    ("Keyboard", 2500, 1),
    ("Mouse", 1200, 2),
    ("USB", 800, 1)
]

totals = []
for name, price, quantity in items:
    totals.append(product_total(name, price, quantity))

sub = subtotal(*totals)
discount = calculate_discount(sub)
after_discount = sub - discount
tax_amount = calculate_tax(after_discount)
final_bill = after_discount + tax_amount

print("\n----- SHOPPING BILL -----")
print("Subtotal: Rs.", sub)
print("Discount: Rs.", discount)
print("Tax: Rs.", tax_amount)
print("Final Bill: Rs.", final_bill)
````

## Question 29 — Student Result Management System

**Problem:**  
Create a complete student result system using functions.

**The system should:**
1. Accept student information
2. Accept multiple subject marks
3. Calculate total
4. Calculate percentage
5. Determine grade
6. Determine pass/fail
7. Display a final result card

**Requirements:**
- Use **multiple functions**
- Use arguments and `return`
- Use lists
- Use conditions and loops
- Use at least one advanced concept (`*args`, `**kwargs`, nested functions, or lambda)
- Do **not** write the entire program inside one function

**Focus:** Combining multiple function concepts

````python
# Write your code below
def get_student_info():
    return {"name": "Hamza", "roll_no": "ST101", "class": "Python"}

def calculate_total(marks):
    return sum(marks)

def calculate_percentage(total, total_marks):
    return (total / total_marks) * 100

def get_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"

def get_status(percentage):
    return "Pass" if percentage >= 40 else "Fail"

def result_card(student, marks):
    total = calculate_total(marks)
    percentage = calculate_percentage(total, len(marks) * 100)
    grade = get_grade(percentage)
    status = get_status(percentage)
    print("----- STUDENT RESULT CARD -----")
    for key, value in student.items():
        print(f"{key.title()}: {value}")
    print("Marks:", marks)
    print("Total:", total, "/", len(marks) * 100)
    print("Percentage:", round(percentage, 2), "%")
    print("Grade:", grade)
    print("Status:", status)

student = get_student_info()
subject_marks = [85, 78, 92, 74, 81]
result_card(student, subject_marks)

# Lambda example: check whether each subject mark is even.
is_even = lambda mark: mark % 2 == 0
print("Even-mark checks:", [is_even(mark) for mark in subject_marks])
````

````python
# Write your code below
def customer_info(name, phone):
    return {"Customer": name, "Phone": phone}

def product_details():
    return [
        {"name": "Keyboard", "price": 2500},
        {"name": "Mouse", "price": 1200},
        {"name": "Headphones", "price": 3000}
    ]

def calculate_item_total(price, quantity):
    return price * quantity

def calculate_subtotal(*amounts):
    return sum(amounts)

def apply_discount(amount, percent=10):
    return amount * percent / 100

def calculate_tax(amount, percent=5):
    return amount * percent / 100

def delivery_charge(subtotal):
    return 0 if subtotal >= 5000 else 250

def make_order(customer, cart):
    item_totals = []
    print("----- ITEMS -----")
    for item in cart:
        quantity = 1
        item_total = calculate_item_total(item["price"], quantity)
        item_totals.append(item_total)
        print(f'{item["name"]} x {quantity}: Rs. {item_total}')

    sub = calculate_subtotal(*item_totals)
    discount = apply_discount(sub)
    after_discount = sub - discount
    tax_amount = calculate_tax(after_discount)
    delivery = delivery_charge(sub)

    # Nested function for the final payable amount.
    def final_amount():
        return after_discount + tax_amount + delivery

    print("\n----- ORDER SUMMARY -----")
    for key, value in customer.items():
        print(f"{key}: {value}")
    print("Subtotal: Rs.", sub)
    print("Discount (10%): Rs.", discount)
    print("Tax (5%): Rs.", tax_amount)
    print("Delivery: Rs.", delivery)
    print("Final Amount: Rs.", final_amount())

    # Lambda example: format each product name in uppercase.
    format_name = lambda product: product["name"].upper()
    print("Products:", [format_name(item) for item in cart])

customer = customer_info("Hamza", "03001234567")
cart = product_details()
make_order(customer, cart)
````
